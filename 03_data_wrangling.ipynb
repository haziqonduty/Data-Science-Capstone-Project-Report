# 03 — Data Wrangling & Processing

Cleans the combined launch table and builds the binary `Class` label used
for modeling. Runs fully offline against `falcon9_dataset.csv`.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../falcon9_dataset.csv")
df.shape
```

### Handle missing values

Payload mass: impute with the column mean.
LandingPad-type fields: leave missing as-is — a missing pad can genuinely mean an ocean landing or no recovery attempt, not bad data.

```python
df["PayloadMass"] = df["PayloadMass"].fillna(df["PayloadMass"].mean())
df.isnull().sum()
```

### Build the binary landing-success label

```python
df["Class"] = df["Outcome"].apply(lambda o: 1 if str(o).startswith("True") else 0)
df["Class"].value_counts()
```

### One-hot encode categorical features for modeling

```python
features = ["LaunchSite", "Orbit", "PayloadMass", "GridFins", "Reused", "Legs", "Block", "FlightNumber"]
X = pd.get_dummies(df[features], columns=["LaunchSite", "Orbit"])
print(X.shape)
X.head()
```

### Save the cleaned, feature-engineered dataset

```python
df.to_csv("../falcon9_clean.csv", index=False)
X.to_csv("../falcon9_features.csv", index=False)
```
