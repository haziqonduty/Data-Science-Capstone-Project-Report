# 04 — EDA with Data Visualization

Generates the exploratory charts used in the presentation slides. Runs
fully offline against `falcon9_dataset.csv`.

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

df = pd.read_csv("../falcon9_dataset.csv")
CYAN, ORANGE = "#14b8a6", "#f59e0b"
```

### Flight number vs payload mass, colored by landing outcome

```python
fig, ax = plt.subplots(figsize=(6,4))
colors = df["Class"].map({1: CYAN, 0: "#94a3b8"})
ax.scatter(df.FlightNumber, df.PayloadMass, c=colors, s=45, edgecolor="white")
ax.legend(handles=[mpatches.Patch(color=CYAN,label="Landed"), mpatches.Patch(color="#94a3b8",label="Not landed")])
ax.set_xlabel("Flight number"); ax.set_ylabel("Payload mass (kg)")
ax.set_title("Flight Number vs Payload Mass")
plt.show()
```

### Launch distribution by orbit

```python
order = df.Orbit.value_counts()
fig, ax = plt.subplots(figsize=(6,4))
ax.bar(order.index, order.values, color=CYAN)
ax.set_ylabel("Launches"); ax.set_title("Launch Distribution by Orbit")
plt.xticks(rotation=45, ha="right")
plt.show()
```

### Launches by site

```python
site_counts = df.LaunchSite.value_counts()
fig, ax = plt.subplots(figsize=(5,4))
ax.bar(site_counts.index, site_counts.values, color=[CYAN, ORANGE, "#818cf8"])
ax.set_ylabel("Launches"); ax.set_title("Number of Falcon 9 Launches by Launch Site")
plt.xticks(rotation=20, ha="right")
plt.show()
```

### Yearly landing-success trend

```python
yearly = df.groupby("Year").Class.mean() * 100
fig, ax = plt.subplots(figsize=(5,4))
ax.plot(yearly.index, yearly.values, marker="o", color=CYAN, linewidth=2)
ax.set_ylabel("Landing success rate (%)"); ax.set_title("Launch Success Yearly Trend")
plt.show()
```

### Success rate by orbit

```python
rate = df.groupby("Orbit").Class.mean().sort_values(ascending=False) * 100
fig, ax = plt.subplots(figsize=(6,4))
ax.bar(rate.index, rate.values, color=CYAN)
ax.set_ylabel("Success rate (%)"); ax.set_title("First-Stage Landing Success Rate by Orbit")
plt.xticks(rotation=45, ha="right")
plt.show()
```
