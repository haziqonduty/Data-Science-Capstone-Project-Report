# 05 — EDA with SQL

Loads the launch table into SQLite and runs the analytical queries behind
the "EDA with SQL" and "SQL Results" slides. Runs fully offline.

```python
import sqlite3
import pandas as pd

df = pd.read_csv("../falcon9_dataset.csv")
conn = sqlite3.connect(":memory:")
df.to_sql("SPACEXTABLE", conn, index=False, if_exists="replace")
cur = conn.cursor()
```

### Distinct launch sites

```python
cur.execute("SELECT DISTINCT LaunchSite FROM SPACEXTABLE;")
cur.fetchall()
```

### Total payload mass for NASA (CRS) missions

```python
cur.execute("""
SELECT SUM(PayloadMass) FROM SPACEXTABLE
WHERE Customer = 'NASA (CRS)';
""")
cur.fetchall()
```

### Average payload mass, Block 5 boosters

```python
cur.execute("SELECT AVG(PayloadMass) FROM SPACEXTABLE WHERE Block = 5;")
cur.fetchall()
```

### Date of the first successful landing on record

```python
cur.execute("SELECT MIN(Date) FROM SPACEXTABLE WHERE Outcome LIKE 'True%';")
cur.fetchall()
```

### Boosters with a successful drone-ship landing, payload 4,000-6,000 kg

```python
cur.execute("""
SELECT FlightNumber, PayloadMass FROM SPACEXTABLE
WHERE Outcome = 'True ASDS' AND PayloadMass BETWEEN 4000 AND 6000;
""")
cur.fetchall()
```

### Total successful vs. failed landing outcomes

```python
cur.execute("""
SELECT CASE WHEN Outcome LIKE 'True%' THEN 'Success' ELSE 'Failure' END AS result,
       COUNT(*) FROM SPACEXTABLE GROUP BY result;
""")
cur.fetchall()
```

### Launch count by year, ranked (time-based analysis)

```python
cur.execute("""
SELECT Year, COUNT(*) as launches FROM SPACEXTABLE
GROUP BY Year ORDER BY launches DESC;
""")
cur.fetchall()
```
